```python
#Import Packages
import pandas as pd
import numpy as np
```

A **Series** is a one-dimensional array

- Unlike a NumPy array, they have an ***index***

```python
#Create a numpy array
arr = np.random.randn(4) #Creates an array of random values from a std normal distribution
print(type(arr))
print(arr, "/n")

#Creat a pandas series from the numpy array
s = pd.Series(arr)
print(type(s))
print(s)
```

- numpy arrays have an implicit index (they can still be indexed by position)
- pandas series have an explicit index (printed as part of the data structure.  Printing a pandas series also shows the data type)

**Creating a pandas series**
s = pd.Series(data, index=index)
- data can be a list, array, dictionary, single value
- index takes a list (optional, but if included it must be as long as the data)

```python
#From a NumPy array
pd.Series(np.arange(3), index = [2023, 2024, 2025])
```

```python
#From a list
#Use the default index
pd.Series(['EDS 220', 'EDS 222', 'EDS 223', 'EDS 242'])
```

```python
#From a dictionary

#Construct the dictionary
d = {'key_0':2, 'key_1':'3', 'key_2':5}

#Initialize the series
pd.Series(d)
```

Data type is an object becuase it is a combination of strings and numbers

```python
#If we give only a single value, the series is filled with that value
pd.Series(3.0, index = ['A', 'B', 'C'])
```

```python
#You can do simple operations on a series

#Define the Series
s = pd.Series([98,73,65], index =['Andrea', 'Beth', 'Carolina'])

#Divide each element by 10
print(s/10, '\n')

#Take the exponential of each element (What is an exponential?)
print(np.exp(s), '\n')

#Unless you save an operation under the same name, the original series is unchanged
print(s)
```

```python
#Produce a new series of T/F indicating whether each value in the original series satisfies a condition
s > 70
```

```python
#We can identify missing numbers

#Make a series with some missing values
s = pd.Series([1, 2, np.nan, 4, np.nan])
print(s)
```

Notice that the data type is still a float-- this is because pandas recognizes NaN as empty values

```python
#Check if there are any empty values or NAs using hasnans
s.hasnans
```

```python
#Use isna to locate them by creating a series of boolean outputs for each value
s.isna()
```

```python
#A dataframe is essentailly a dictionary of pandas Series.  Create it like one!

#Innitalize the dictionary with the data for each column
d = {'col_name_1':pd.Series(np.arange(3)),
     'col_name_2':pd.Series([3.1, 3.2, 3.3])
     }

#Create the dataframe
df = pd.DataFrame(d)
df
```
